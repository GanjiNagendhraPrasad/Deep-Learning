```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('./breast-cancer.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
df=df.drop(['id','Unnamed: 32'],axis=1)
df.head()
```

```python
df["diagnosis"]=df["diagnosis"].map({'B':0,'M':1}).astype(int)
df.head()
```

```python
X = df.iloc[: , 1:]
y = df.iloc[: , 0]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
X_train.head()
```

```python
y_train.head()
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
import tensorflow
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Dropout
from tensorflow.keras.layers import ReLU,LeakyReLU,PReLU,ELU
```

```python
ann = Sequential()
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
X.shape
```

```python
ann.add(Dense(units=6,kernel_initializer='he_uniform',activation='relu',input_dim=X.shape[1]))
ann.add(Dense(units=6,kernel_initializer='he_uniform',activation='relu'))
ann.add(Dense(units=1,kernel_initializer='glorot_uniform',activation='sigmoid'))

```

```python
ann.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
reg = ann.fit(X_train,y_train,validation_split=0.1,batch_size=10,epochs=50)
```

```python
y_pred = ann.predict(X_test)
y_pred = y_pred > 0.5
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
```

```python
print(confusion_matrix(y_test,y_pred))
```

```python
print(classification_report(y_test,y_pred))
```

```python
accuracy_score(y_test,y_pred)
```
